```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
from pathlib import Path


from sklearn.datasets import fetch_california_housing
```

```python


# Load the dataset
housing = fetch_california_housing()

# Create DataFrame
df = pd.DataFrame(
    housing.data,
    columns=housing.feature_names
)

# Add target column
df["MedHouseVal"] = housing.target

# Explore the dataset
df.head()

df.shape #tells us the number of rows and columns(size) in the dataset

df.columns

df.dtypes

df.info() #gives information about the dataset

df.describe() #gives a statistical summary.

df.isnull().sum() #counts the number of missing values in each column.
```

```python
#EDA
plt.figure(figsize=(8,5))

plt.hist(df["MedInc"], bins=30, edgecolor='k')

plt.title("Distribution of Median Income")
plt.xlabel("Median Income")
plt.ylabel("Number of Houses")

plt.show()
```

```python
#Outlier detection

plt.figure(figsize=(8,4))

plt.boxplot(df["MedInc"])
plt.title("Median Income Boxplot")

plt.show()

#Correlation analysis
correlation_matrix =df.corr(numeric_only=True)

correlation_matrix

```

```python
#Heatmap of the correlation matrix
plt.figure(figsize=(10, 8))

sns.heatmap(correlation_matrix,
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Matrix of California Housing Dataset")

plt.show()
```

```python
#Scatter plot of Median Income vs Median House Value
plt.figure(figsize=(8,5))

plt.scatter(df['MedInc'], 
            df['MedHouseVal'])

plt.title("Income vs House Value")

plt.show()
```

```python
#Train-test split


#Features
X = df.drop("MedHouseVal", axis=1)

#Target
y = df["MedHouseVal"]

#X (uppercase) contains all the features.
#y (lowercase) contains the target variable.

#Split the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y, 
    test_size=0.2, #means 20% of the data will be used for testing and 80% for training.
    random_state=42) #randomly shuffles the data before splitting to ensure that the training and testing sets are representative of the overall dataset.

print(X_train.shape)
print(X_test.shape)
print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

```python
#Linear Regression Model-training the model and making predictions

model = LinearRegression()

#we're telling the model:
#"Study these examples and learn the relationship between the features and the house price."
#fit() is the learning phase.
model.fit(X_train, y_train) #fit the model to the training data

#predict() is the inference phase
predictions = model.predict(X_test) #make predictions on the test data

# Display the first five predictions
print(predictions[:5])
```

```python
#Evaluating the model

#Calculate the Mean Absolute Error (MAE) and R-squared (R²) score to evaluate the model's performance.

mae= mean_absolute_error(y_test, predictions)
print(mae)

```

```python
#Calculating R score

r2 = r2_score(y_test, predictions)

print(r2)

#got 0.57 output, that means the model explains 57% of the variance in the target variable (Median House Value) based on the features provided.
```

```python
#Joblib

# Train model
model.fit(X_train, y_train)


# Define the path relative to your working environment
BASE_DIR = Path.cwd()
MODELS_DIR = BASE_DIR / "models"
MODEL_FILE = MODELS_DIR / "house_price_model.pkl"

# Ensure the directory exists
MODELS_DIR.mkdir(exist_ok=True)

# Save model
joblib.dump(model, MODEL_FILE)
print("Model saved successfully!")

# Load model using the exact same path reference
loaded_model = joblib.load(MODEL_FILE)
print("Model loaded successfully!")

# Predict
predictions = loaded_model.predict(X_test)
print(predictions[:5])

```
