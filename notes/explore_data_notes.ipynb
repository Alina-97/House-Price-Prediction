Convert to a dataframe
-----------------------
We convert only if the data is not already in a dataframe format. In this case, we are using the sklearn library to load the data, which returns a Bunch object. We can convert it to a dataframe using the following code:

df = pd.DataFrame(housing.data,
columns=housing.feature_names)

df = pd.read_csv('../data/housing.csv')
df.head(5)

We're telling Pandas:

"Create a table using these numbers and these column names."
