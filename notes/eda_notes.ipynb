Exploratory Data Analysis (EDA)
--------------------------------

It's the process of exploring your dataset to answer questions like:

1. What does the data look like?
2. Is the data clean?
3. Which features are important?
4. Are there any strange values?
5. What patterns exist?

Step 1 - Looking at the Distribution of a Feature
-------------------------------------------------

Step 2 — Boxplots
Outlier
An outlier is a value that is much higher or lower than the rest.

Why Outliers Matter

Some algorithms are sensitive to them.


Step 3 — Correlation

What is Correlation?

Correlation measures how strongly two variables move together.

We have positive and negative correlation

0.00  → No relationship

0.30  → Weak

0.50  → Moderate

0.70+ → Strong

1.00  → Perfect

Calculating Correlation
-------------------------
correlation = df.corr(numeric_only=True)
correlation

This produces a correlation matrix.
