Features
========

Is the information used to make a prediction.

Features describe an object.

For our project:
Bedrooms
Bathrooms
Area
Garage
Age

Think of features as the information you would tell an estate agent before asking for a valuation.

Target
======

Is the value we want to predict also called label

In this case Price is our target

Termilogies Used Constantly

| Term       | Meaning                                  |
| ---------- | ---------------------------------------- |
| Feature    | An input used for prediction             |
| Target     | The value we want to predict             |
| Dataset    | A collection of examples                 |
| Row        | One observation (one house)              |
| Column     | One feature or the target                |
| Model      | A program that learns patterns from data |
| Prediction | The model's estimated output             |



In Machine Learning, we usually try to collect many features because more relevant information often helps the model make better predictions.

Example:

| Feature             | Example                    |
| ------------------- | -------------------------- |
| Years of experience | 5                          |
| Education level     | Bachelor's                 |
| Job title           | Software Developer         |
| Industry            | Banking                    |
| Company size        | Enterprise                 |
| Location            | Johannesburg               |
| Skills              | Java, Flutter, Spring Boot |
| Certifications      | AWS, Azure                 |
| Age (sometimes)     | 29                         |

Target:
Annual Salary

salary is never used as a feature because that's what we're trying to predict

IMPORTANT RULE

Features

These are the pieces of information you already know before making a prediction

Target

This is the information you do not know and want the computer to predict.


The model learns a mapping like this:
Features
    │
    ▼
Machine Learning Model
    │
    ▼
Target



You cannot train a supervised machine learning model without an answer key 


Core of Supervised Learning

The model learns from examples where the correct answers are already known.

Unsupervised Learning

finding patterns without knowing the answers

| Supervised Learning           | Unsupervised Learning       |
| ----------------------------- | --------------------------- |
| Correct answers are provided  | No correct answers          |
| Learns input → output mapping | Finds hidden patterns       |
| Predicts a target             | Discovers structure in data |
