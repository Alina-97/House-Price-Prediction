Evaluating ML model
===================

Mean Absolute Error (MAE)
Mean=Average
Absolute=Ignore whether the error is positive or negative
Error=Difference between prediction and reality


MAE means the average distance between our predictions and the real values

We use the absolute value because we're interested in how big the mistake is

Example:
Prediction A

Actual: R500,000
Predicted: R490,000
Error = -R10,000

Prediction B

Actual: R500,000
Predicted: R510,000
Error = +R10,000


Mean Squared Error (MSE)
=========================
Instead of taking the absolute value, MSE squares each error.

When we square a number, the result is always non-negative.

Calculating the mean:
1. Add them together.
2. Now divide by the number of errors.

Root Mean Squared Error (RMSE)
==============================

RMSE is simply:
The square root of MSE.

Example:
MSE = 4.67
↓
RMSE = √4.67
↓
≈ 2.16

When Should We Use Each?
========================
| Metric   | Best Use                                                                               |
| -------- | -------------------------------------------------------------------------------------- |
| **MAE**  | Easy to explain. Treats all errors equally.                                            |
| **MSE**  | Penalizes large mistakes much more heavily. Common in optimization and model training. |
| **RMSE** | Same idea as MSE but easier to interpret because it's in the original unit.            |


RMSE punishes large mistakes much more than small ones.

What Does This Tell Us?
A high RMSE compared to MAE usually means:
The model occasionally makes very large mistakes.


R² Score (Coefficient of Determination)
=======================================
Unlike MAE and RMSE, R² doesn't measure error directly.

Instead, it answers this question:
How much of the variation in house prices can my model explain?

Interpreting R²

|    R² Score | Meaning                                          |
| ----------: | ------------------------------------------------ |
|        1.00 | Perfect predictions                              |
|        0.95 | Excellent model                                  |
|        0.80 | Strong model                                     |
|        0.60 | Reasonable model                                 |
|        0.00 | No better than predicting the average every time |
| Less than 0 | Worse than simply predicting the average         |


 higher is better for R².

This is the opposite of MAE and RMSE, where lower is better.

Putting It All Together
=======================

Imagine you train a model and get:
MAE  = R18,500
RMSE = R24,200
R²   = 0.87

As an ML engineer, you'd summarize it like this:

"On average, our predictions are about R18,500 away from the actual house prices. The model doesn't make many large errors, and it explains 87% of the variation in house prices.
